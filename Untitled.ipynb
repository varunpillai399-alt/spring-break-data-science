```python
prices = [180.0, 182.5, 178.3, 185.0, 183.2]

mean = sum(prices) / len(prices)
print(round(mean, 2))
```

```python
prices = [180.0, 182.5, 178.3, 185.0, 183.2]

mean = sum(prices) / len(prices)

deviations = [(p - mean) ** 2 for p in prices]  # list comprehension!
variance = sum(deviations) / len(deviations)

print(round(variance, 2))  # 5.17
```

```python
prices = [180.0, 182.5, 178.3, 185.0, 183.2]

min_p = min(prices)
max_p = max(prices)

normalized = [(p - min_p) / (max_p - min_p) for p in prices]

print([round(n, 2) for n in normalized])
# [0.26, 0.63, 0.0, 1.0, 0.72]
```
